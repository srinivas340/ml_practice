```python
import matplotlib.pyplot as plt
```

```python
import numpy as np
```

```python
x=np.linspace(0,2*np.pi,100)
print(x[-1],2*np.pi)
print(x)
```

```python
y=np.sin(x)
z=np.cos(x)
w=np.sin(4*x)
print(y)
print(z)
print(w)
```

```python
plt.plot(x,y)
plt.xlabel('x')
plt.ylabel('y(x)')
plt.show()
```

```python

plt.plot(x,z)
plt.xlabel('x')
plt.ylabel('y(x)')
plt.show()
```

```python
plt.plot(x,w)
plt.xlabel('x')
plt.ylabel('y(x)')
plt.show()
```

```python
#parabola
x=np.linspace(-10,10,20)
y=x**2
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,y,"r+")
```

```python
plt.plot(x,y,"r--")
```

```python
x=np.linspace(-5,5,50)
plt.plot(x,np.sin(x))
plt.plot(x,np.cos(x))
plt.xlabel('x')
plt.ylabel('y(x)')
plt.show()
```

bar plot

```python
fig=plt.figure()
ax=fig.add_axes([0,0,1,1])
x=['a','b','c']
y=[3999,499,59999]
ax.bar(x,y)
plt.show()
```

pie chat

```python
fig1=plt.figure()
ax1=fig1.add_axes([0,0,1,1])
labels=['a','b','c']
ax1.pie(y,labels=labels,autopct='%2.1f%%')
ax1.axis('equal')
plt.show()
```

Scatter plot

```python
x=np.linspace(0,10,20)
y=np.random.rand(20)
plt.scatter(x,y)
plt.show()
```

3d scatter plot

```python
fig3=plt.figure()
ax3=plt.axes(projection='3d')
z=np.random.random(100)
x=np.random.random(100)
y=np.random.random(100)
ax3.scatter3D(x,y,z)
plt.show()
```
