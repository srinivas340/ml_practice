```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

Seaborn has some uilt in datasets

```python
#total bill vs tip dataset
tips=sns.load_dataset('tips')
tips.head()
```

```python
#setting theam for plot
sns.set_theme()
```

```python
#visuallize the data
sns.relplot(x='total_bill',y='tip',data=tips,hue='sex',style='smoker',size='size')
```

```python
#load iris dataset
iris=sns.load_dataset('iris')
iris.head()
```

Scatter plot

```python
sns.scatterplot(x='sepal_length',y='sepal_width',hue='species',data=iris)
```

```python
sns.scatterplot(x='petal_length',y='petal_width',hue='species',data=iris)
```

```python
#loading titanic data set
titanic=sns.load_dataset('titanic')
titanic.head()
```

Count plot

```python
sns.countplot(x='sex',data=titanic)
```

```python
sns.countplot(x='class',data=titanic)
```

```python
sns.countplot(x='who',data=titanic)
```

Bar chart

```python
sns.barplot(x='sex',y='survived',hue='class',data=titanic)
```

```python
#house price dataset
from sklearn.datasets import fetch_california_housing
house=fetch_california_housing()
house=pd.DataFrame(house.data,columns=house.feature_names)
house.head()
```

Distribution plot

```python
sns.distplot(house['AveRooms'])
```

```python
sns.distplot(house['Population'])
```

Correlation

1.positive correlation

2.negitive correlation

Heat Map

```python
plt.figure(figsize=(10,10))
sns.heatmap(house.corr(),annot=True)
```
