Imbalanced dataset:

A dataset with an unequal class distribution

```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('/content/credit_data.csv')
```

```python
df.head()
```

```python
df.shape
```

```python
df['Class'].value_counts()
```

0-->Legt Transactions

1-->Fraudulent Transaction

```python
legit=df[df.Class==0]
fraud=df[df.Class==1]
```

```python
print(legit.shape)
print(fraud.shape)
```

Under sampling

Build a sample dataset containing similar distribution of legit & fraudulent transactions



```python
legit_sample=legit.sample(n=211)
```

```python
legit_sample.shape
```

Concatenate the two dataframe

```python
new_df=pd.concat([legit_sample,fraud],axis=0)
```

```python
new_df.head()
```
